<h3>Step 1: Load and Explore the Dataset</h3>

Explore how respondents are distributed across different industries.

```python
##Write your code here
industry_counts = df["Industry"].value_counts().head(15)  # top 15 industries

# --- Step 2: Plot bar chart ---
plt.figure(figsize=(10,6))
industry_counts.plot(kind="bar", color="teal", edgecolor="black")

plt.title("Distribution of Respondents by Industry")
plt.xlabel("Industry")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# --- Step 3: Print top industries ---
print("Top Industries by Respondent Count:")
print(industry_counts)
```

<h3>Step 3: Identify High Compensation Outliers</h3>


Identify respondents with extremely high yearly compensation.

- Calculate basic statistics (mean, median, and standard deviation) for `ConvertedCompYearly`.

- Identify compensation values exceeding a defined threshold (e.g., 3 standard deviations above the mean).


```python
##Write your code here
comp = df["ConvertedCompYearly"].dropna()

# Step 2: Basic statistics
mean_val = comp.mean()
median_val = comp.median()
std_val = comp.std()

print("Mean yearly compensation:", mean_val)
print("Median yearly compensation:", median_val)
print("Standard Deviation:", std_val)

# Step 3: Define outlier threshold
threshold = mean_val + (3 * std_val)
print("Outlier Threshold (mean + 3*std):", threshold)

# Step 4: Identify outliers
outliers = df[df["ConvertedCompYearly"] > threshold][["ConvertedCompYearly"]]

print("\nNumber of outliers:", len(outliers))
print("Top 10 outliers:")
print(outliers.sort_values(by="ConvertedCompYearly", ascending=False).head(10))
```

<h3>Step 4: Detect Outliers in Compensation</h3>


Identify outliers in the `ConvertedCompYearly` column using the IQR method.

- Calculate the Interquartile Range (IQR).

- Determine the upper and lower bounds for outliers.

- Count and visualize outliers using a box plot.


```python
Q1 = comp.quantile(0.25)
Q3 = comp.quantile(0.75)
IQR = Q3 - Q1

print("Q1 (25th percentile):", Q1)
print("Q3 (75th percentile):", Q3)
print("IQR:", IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = comp[(comp < lower_bound) | (comp > upper_bound)]
print("Number of outliers:", len(outliers))

plt.figure(figsize=(8,6))
plt.boxplot(comp, vert=False, patch_artist=True, boxprops=dict(facecolor="skyblue"))
plt.title("Boxplot of ConvertedCompYearly with Outliers")
plt.xlabel("Yearly Compensation")
plt.show()
```

<h3>Step 5: Remove Outliers and Create a New DataFrame</h3>


Remove outliers from the dataset.

- Create a new DataFrame excluding rows with outliers in `ConvertedCompYearly`.
- Validate the size of the new DataFrame.


```python
##Write your code here
df_no_outliers = df[(df['ConvertedCompYearly'] >= lower_bound) & 
                    (df['ConvertedCompYearly'] <= upper_bound)]

print("Original DataFrame size:", df.shape)
print("New DataFrame size without outliers:", df_no_outliers.shape)
```
