### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# Check unique entries in Country column
print(df["Country"].unique())
# Frequency counts
print(df["Country"].value_counts(dropna=False).head(20))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
print("Unique Country values (sample):")
print(df["Country"].unique()[:20])   # first 20 values
print("\nUnique EdLevel values (sample):")
print(df["EdLevel"].unique()[:20])   # first 20 values

# --- Step 2: Define mapping dictionaries ---

# Country mapping
country_map = {
    "US": "United States",
    "U.S.A.": "United States",
    "USA": "United States",
    "United States of America": "United States",
    "UK": "United Kingdom",
    "England": "United Kingdom",
    "Scotland": "United Kingdom",
    "Republic of India": "India",
    "Korea, South": "South Korea"
}

# Education Level mapping
edlevel_map = {
    "Bachelor's degree": "Bachelors",
    "Bachelors": "Bachelors",
    "Bachelor degree": "Bachelors",
    "Master’s degree": "Masters",
    "Masters": "Masters",
    "PhD": "Doctorate",
    "Doctor of Philosophy": "Doctorate",
    "Some college": "Undergraduate",
    "Some College": "Undergraduate",
    "High school": "Secondary",
    "Secondary school": "Secondary"
}
# --- Step 3: Apply the mappings ---
df["Country"] = df["Country"].replace(country_map)
df["EdLevel"] = df["EdLevel"].replace(edlevel_map)

# --- Step 4: Remove irrelevant or placeholder values (if present) ---
irrelevant_countries = ["Other", "Unknown", "-", "N/A", "?", "Prefer not to say"]
df = df[~df["Country"].isin(irrelevant_countries)]

irrelevant_ed = ["Other", "Unknown", "-", "N/A", "?", "Prefer not to say"]
df = df[~df["EdLevel"].isin(irrelevant_ed)]

# --- Step 5: Verify results ---
print("\nStandardized Country values (top 10):")
print(df["Country"].value_counts().head(10))

print("\nStandardized Education Levels:")
print(df["EdLevel"].value_counts())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
import pandas as pd

# Load dataset
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(url)

# --- Step 1: Define mapping function ---
def simplify_employment(value):
    if pd.isna(value):
        return "Missing"
    val = value.lower()
    if "full-time" in val:
        return "Full-time"
    elif "part-time" in val:
        return "Part-time"
    elif "student" in val and "full-time" not in val and "part-time" not in val:
        return "Just Student"
    elif "retired" in val:
        return "Retired"
    elif "unemployed" in val:
        return "Unemployed"
    else:
        return "Other"

# --- Step 2: Apply mapping ---
df["Employment_Simplified"] = df["Employment"].apply(simplify_employment)

# --- Step 3: One-hot encode the simplified categories ---
employment_encoded = pd.get_dummies(df["Employment_Simplified"], prefix="Employment")

# --- Step 4: Merge back ---
df = pd.concat([df, employment_encoded], axis=1)

# --- Step 5: Verify result ---
print(df[["Employment", "Employment_Simplified"]].head(10))
print("\nOne-hot encoded columns:")
print(employment_encoded.head())

```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
missing_values = df.isna().sum().sort_values(ascending=False)

# Show top 10 columns with most missing values
print("Columns with the highest number of missing values:")
print(missing_values.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
df["ConvertedCompYearly"].fillna(df["ConvertedCompYearly"].mean(), inplace=True)

# Verify
print("Remaining missing values:", df["ConvertedCompYearly"].isna().sum())
```

```python
df["ConvertedCompYearly"].head()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
most_frequent_value = df["RemoteWork"].value_counts().idxmax()
print("Most frequent RemoteWork value:", most_frequent_value)

# Step 2: Impute missing values
df["RemoteWork"].fillna(most_frequent_value, inplace=True)

# Step 3: Verify
print("Remaining missing values in RemoteWork:", df["RemoteWork"].isna().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
df["ConvertedCompYearly_MinMax"] = (
    (df["ConvertedCompYearly"] - df["ConvertedCompYearly"].min()) /
    (df["ConvertedCompYearly"].max() - df["ConvertedCompYearly"].min())
)

# Verify
print(df[["ConvertedCompYearly", "ConvertedCompYearly_MinMax"]].head())
```
